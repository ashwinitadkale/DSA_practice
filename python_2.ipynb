### `Problem 1`: Write a program that will give you in hand monthly salary after deduction on CTC - HRA(10%), DA(5%), PF(3%) and taxes deduction as below:

> Salary(Lakhs) : Tax(%)

*   Below 5 : 0%
*   5-10 : 10%
*   10-20 : 20%
*   aboove 20 : 30%

````python
ctc = float(input("Enter CTC in lakhs: "))

# Convert lakhs to actual amount
ctc_amount = ctc * 100000

# Deductions
hra = 0.10 * ctc_amount
da = 0.05 * ctc_amount
pf = 0.03 * ctc_amount

salary_after_deductions = ctc_amount - (hra + da + pf)

# Tax calculation
if ctc < 5:
    tax_rate = 0
elif 5 <= ctc < 10:
    tax_rate = 0.10
elif 10 <= ctc < 20:
    tax_rate = 0.20
else:
    tax_rate = 0.30

tax = salary_after_deductions * tax_rate

final_yearly_salary = salary_after_deductions - tax
monthly_in_hand = final_yearly_salary / 12

print("Monthly In-hand Salary =", round(monthly_in_hand, 2))
````

### `Problem 6` - Find the factorial of a given number.

Write a program to use the loop to find the factorial of a given number.

The factorial (symbol: `!`) means to multiply all whole numbers from the chosen number down to 1.

For example: calculate the factorial of 5

```bash
5! = 5 × 4 × 3 × 2 × 1 = 120
```

Output:

```bash
120
```

````python
n=int(input("Enter a number: "))
if n<0:
  print("Factorial does not exist for negative numbers.")
elif n==0:
  print("Factorial =1")
else:
  fact=1
  for i in range(1,n+1):
    fact*=i
  print("Factorial=",fact)
````
