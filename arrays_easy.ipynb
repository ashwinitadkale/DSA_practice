# 1.Find the Largest element in an array

```python
def largest(arr):
  maxi=arr[0]
  for num in arr:
      if num>maxi:
          maxi=num
  return maxi
arr=[1,2,3,4,5]
print("largest element is :",largest(arr))
```

# 2.Find Second Smallest and Second Largest Element in an array

## 2.1 Brute Force

Time Complexity: O(N log N), for sorting the array.

Space Complexity: O(1), as we are using a constant amount of space for variables.

```python
def brute_second(arr,n):
  if n==0 or n==1:
    print(-1,-1)
  arr.sort()
  return arr[n-2],arr[1]
a=[4,5,67,8,9,6]
print("second largest and second smallest elements are:",brute_second(a,len(a)))
```

## 2.2 Better Approach

Time Complexity: O(N), we do two linear traversals in our array.

Space Complexity: O(1), as we are using a constant amount of space for variables.

```python
def better_second(arr,n):
  if n==0 or n==1:
    print(-1,-1)
    return
  small=float('inf')
  second_small=float('inf')
  large=float('-inf')
  second_large=float('-inf')
  for i in range(n):
    small=min(small,arr[i])
    large=max(large,arr[i])
  for i in range(n):
    if arr[i]<second_small and arr[i]!=small:
      second_small=arr[i]
    if arr[i]>second_large and arr[i]!=large:
      second_large=arr[i]
  print("second smallest is",second_small)
  print("second largest is",second_large)
a=[1,2,3,4,5,6,7,8]
n=8
better_second(a,n)

```

## 2.3 Optimal Approach

Time Complexity: O(N), we do two linear traversals in our array.

Space Complexity: O(1), as we are using a constan

```python
def second_smallest(arr,n):
  if n<2:
    return -1
  small=float('inf')
  second_small=float('inf')
  for i in range(n):
    if arr[i]<small:
      second_small=small
      small=arr[i]
    elif arr[i]<second_small and arr[i]!=small:
      second_small=arr[i]
  return second_small
def second_largest(arr,n):
  if n<2:
    return -1
  large=float('-inf')
  second_large=float('-inf')
  for i in range(n):
    if arr[i]>large:
      second_large=large
      large=arr[i]
    elif arr[i]>second_large and arr[i]!=large:
      second_large=arr[i]
  return second_large
arr=[1,2,4,7,7,5]
n=6
ss=second_smallest(arr,n)
sl=second_largest(arr,n)
print(f"Second smallest is {ss}")
print(f"Second largest is {sl}")



```

# 3.Check if an Array is Sorted

```python
def isSorted(arr,n):
  for i in range(1,n):
    if arr[i]<arr[i-1]:
      return False
  return True
arr = [1, 2, 3, 8, 5]
n = len(arr)
print("True" if isSorted(arr,n) else "False")
```

# 4.Remove Duplicates in-place from Sorted Array

4.1 Brute Force Approach

Time Complexity: O(N), We traverse the entire array and insert elements into set.

Space Complexity: O(N), additional space used to store elements in set.


```python
def remove_dup(arr):
  unique=sorted(set(arr))
  for i in range(len(unique)):
    arr[i]=unique[i]
  return len(unique),list(unique)
arr=[1,1,2,2,3,4,4]
print(remove_dup(arr))
```

4.2 Optimal Approach

Time Complexity: O(N), We traverse the entire original array only once.

Space Complexity: O(1), constant additional space is used to check unique elements.


```python
def remove_dup_op(arr):
  i=0
  for j in range(1,len(arr)):
    if arr[j]!=arr[i]:
      i+=1
      arr[i]=arr[j]
  return i+1
arr=[1,1,2,2,3,4,4]
print(remove_dup_op(arr))
```

# 5.Left Rotate the Array by One

5.1 Brute Force

Time Complexity: O(N), where N is the size of the array. This is because we traverse the array once to shift the elements.

Space Complexity: O(N), as we are using a temporary array of the same size as the input array to store the shifted elements

```python
def left_rotate1(arr,n):
  temp=[0]*n
  for i in range(1,len(arr)):
     temp[i-1]=arr[i]
  temp[-1]=arr[0]
  return temp
arr=[1,2,3,4]
print(left_rotate1(arr,len(arr)))
```

5.2 Optimal Approach

Time Complexity: O(N), where N is the size of the input array. This is because we traverse the array once to shift the elements.

Space Complexity: O(1), as we are using only a constant amount of extra space for the temporary variable.

```python
def left_rotate2(arr,n):
  temp=arr[0]
  for i in range(1,len(arr)):
    arr[i-1]=arr[i]
  arr[-1]=temp
  return arr
arr=[1,2,3,4]
print(left_rotate2(arr,len(arr)))
```

# 6.Left rotate by k

```python
def rotate_k(nums,k):
  n=len(nums)
  k=k%n
  def reverse(start,end):
    while start<end:
      nums[start],nums[end]=nums[end],nums[start]
      start+=1
      end-=1
  reverse(0,n-1)
  reverse(0,k-1)
  reverse(k,n-1)
  return nums
nums=[1,2,3,4,5,6,7]
k=3
print(rotate_k(nums,k))
```

# 7.Move all Zeroes to the end

7.1 Brute Force Method

Time Complexity: O(N), we can move all zeroes to end in linear time.

Space Complexity: O(N), additional space used for temporary array.

```python
def move_zeros(arr):
    temp = []

    # Step 1: store non-zero elements
    for num in arr:
        if num != 0:
            temp.append(num)

    # Step 2: copy back
    for i in range(len(temp)):
        arr[i] = temp[i]

    # Step 3: fill zeros
    for i in range(len(temp), len(arr)):
        arr[i] = 0

    return arr
arr=[0,1,0,3,12]
print(move_zeros(arr))
```

7.2 Optimal Approach

Time Complexity: O(N), we can move all zeroes to end in linear time.

Space Complexity: O(1), constant additional space is used.


```python
def move_zeroes(nums):
  j=0
  for i in range(len(nums)):
    if nums[i]!=0:
      nums[i],nums[j]=nums[j],nums[i]
      j+=1
  return nums
nums=[0,1,0,3,12]
print(move_zeroes(nums))
```

# 8. Linear Search

Time Complexity: O(N), where N is the number of elements in the array. This is because we traverse the entire array to find the element.

Space Complexity: O(1), as we are using a constant

```python
def linear_search(nums,k):
  for i in range(len(nums)):
    if nums[i]==k:
      return i
nums=[1,2,4,3,5,6]
print(linear_search(nums,6))
```

# 9. Find the number that appears once, and the other numbers twice

9.1 Brute Force
Time Complexity: O(N*N), since nested for loops are used

Space Complexity: O(1). No extra space used

```python
def find_num_brute(nums):
  for i in range(len(nums)):
    count=0
    for j in range(len(nums)):
      if nums[i] == nums[j]:
        count += 1

    if count == 1:
      return nums[i]
nums=[4,1,2,1,2]
print(find_num_brute(nums))
```
