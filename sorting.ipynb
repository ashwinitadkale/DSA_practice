Merge Sort

```python
nums=[10,9,1,1,1,2,3,1]
def merge_sort(nums):
    if len(nums)<=1:
        return nums
    mid=len(nums)//2
    left=merge_sort(nums[:mid])
    right=merge_sort(nums[mid:])
    return merge(left,right)
def merge(left,right):
    res=[]
    i=0
    j=0
    while i<len(left) and j<len(right):
        if left[i]<right[j]:
            res.append(left[i])
            i+=1
        else:
            res.append(right[j])
    res.extend(left[i:])
    res.extend(right[j:])
    return res
```

Selection Sort

```python
#time comp O(n2) best,avg and worst all same
def selectionSort(arr):
    n=len(arr)
    for i in range(n-1):
        mini=i
        for j in range(i+1,n):
            if arr[j]<arr[mini]:
                mini=j
        arr[i],arr[mini]=arr[mini],arr[i]
    return arr
arr1=[13,46,24,52,20,9]
print(selectionSort(arr1))
```

Bubble sort

```python
#time comp best: O(n) avg: O(n2) worst O(n2) space: O(1)
def bubbleSort(arr):  
    n=len(arr)   
    for i in range(n):   
        swapped=False  
        for j in range(0,n-i-1):  
            if arr[j]>arr[j+1]:    
                arr[j],arr[j+1]=arr[j+1],arr[j]   
                swapped=True
        if not swapped:
            break
    return arr
arr1=[13,46,24,52,20,9] 
print(bubbleSort(arr1))
```

```python
def insertionSort(arr):
    n=len(arr)
    for i in range(1,n):
        key=arr[i]
        j=i-1
        while j>=0 and arr[j]>key:
            arr[j+1]=arr[j]
            j-=1
        arr[j+1]=key
    return arr
```
