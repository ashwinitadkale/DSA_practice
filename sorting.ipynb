Bubble Sort 

```python
nums=[10,9,1,1,1,2,3,1]
n=len(nums)
for i in range(n):
    for j in range(0,n-i-1):
        if nums[j]>nums[j+1]:
            nums[j],nums[j+1]=nums[j+1],nums[j]
    print(nums)
    break
```

Merge Sort

```python
nums=[10,9,1,1,1,2,3,1]
def merge_sort(nums):
    if len(nums)<=1:
        return nums
    mid=len(nums)//2
    left=merge_sort(nums[:mid])
    right=merge_sort(nums[mid:])
    return merge(left,right)
def merge(left,right):
    res=[]
    i=0
    j=0
    while i<len(left) and j<len(right):
        if left[i]<right[j]:
            res.append(left[i])
            i+=1
        else:
            res.append(right[j])
    res.extend(left[i:])
    res.extend(right[j:])
    return res
```
