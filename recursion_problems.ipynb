Implement Pow(x,n) | X raised to the power N

```python
#Brute force
#Time Complexity: O(n), where n is the absolute value of the exponent. This is because we multiply the base x, n times.
#Space Complexity: O(1), as we are using a constant amount of space for the variables used in the computation.
def pow(x,n):
    if n==0 or x==1:
        return 1
    temp=n
    if n<0:
        x=1/x
        temp=-1*n
    ans=1
    for i in range(temp):
        ans*=x
    return ans

print(pow(2,3))
print(pow(2,-3))
```

```python
# Optimal Approach
# Time Complexity: O(log n), where n is the absolute value of the exponent. This is because we reduce the problem size by half in each recursive call when n is even.
# Space Complexity: O(log n), due to the recursive call stack. In the worst case, the depth of the recursion can go up to log(n) when n is even.
def myPow(x,n):
    def fast_pow(x,n):
        if n==0:
            return 1
        if n%2==0:
            return fast_pow(x*x,x//2)
        else:
            return x*fast_pow(x,n-1)
    if n<0:
        return 1/fast_pow(x,-n)
    return fast_pow(x,n)

print(myPow(2,3))
print(myPow(2,-3))        
```

Count Good numbers

```python

def countGoodNumbers(n):
    MOD = 10**9 + 7
    def power(x, n):
        if n == 0:
            return 1
            
        half = power(x, n // 2) % MOD
            
        if n % 2 == 0:
            return (half * half) % MOD
        else:
            return (half * half * x) % MOD
        
    even = (n + 1) // 2
    odd = n // 2
        
    return (power(5, even) * power(4, odd)) % MOD
print(countGoodNumbers(3))
```

Sort a Stack

```python
def insert(stack,temp):
    if not stack or stack[-1]<=temp:
        stack.append(temp)
        return
    val=stack.pop()
    insert(stack,temp)
    stack.append(val)
def sortStack(stack):
    if stack:
        temp=stack.pop()
        sortStack(stack)
        insert(stack,temp)
s=[4,1,3,2]
sortStack(s)
print(s)
```

```python
def insertAtBottom(stack,temp):
    if not stack:
        stack.append(temp)
        return
    top=stack.pop()
    insertAtBottom(stack,temp)
    stack.append(top)
def revStack(stack):
    if not stack:
        return
    if stack:
        temp=stack.pop()
        revStack(stack)
        insertAtBottom(stack,temp)
s=[4,1,3,2]
revStack(s)
print(s)
```
