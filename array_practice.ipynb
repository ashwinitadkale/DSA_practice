Largest element in the array

```python
#Brute Force
arr=[1,2,3,4,5,2,1]
arr.sort()
print(arr[-1])
```

```python
#Optimal Approach
arr=[1,2,3,4,5,2,1]
maxi=arr[0]
for i in range(len(arr)):
    if arr[i]>maxi:
        maxi=arr[i]
print(maxi)
```
