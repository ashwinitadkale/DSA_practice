Merge Two strings

```python
#Brute Force #time complexity o(n^2) space complexity o(n+m)
def mergeAlternately(word1,word2):
    result=""
    i=0
    j=0
    while i<len(word1) and j<len(word2):
        result+=word1[i]
        result+=word2[j]
        i+=1
        j+=1
    while i<len(word1):
        result+=word1[i]
        i+=1
    while j<len(word2):
        result+=word2[j]
        j+=1
    return result
print(mergeAlternately('abcd','tuvxyz'))
```

```python
#Optimal Approach
def mergeAlternately(word1,word2):
    result=[]
    i=0
    j=0
    while i<len(word1) and j<len(word2):
        result.append(word1[i])
        result.append(word2[j])
        i+=1
        j+=1
    while i<len(word1):
        result.append(word1[i])
        i+=1
    while j<len(word2):
        result.append(word2[j])
        j+=1
    return ''.join(result)
print(mergeAlternately('abcd','tuvxyz'))
```

Merge Sorted Arrays

```python
#Brute Force
def merge(num1,m,num2,n):
    for i in range(n):
        num1[m+i]=num2[i]
    num1.sort()
    return num1
print(merge([1,2,3,0,0,0],3,[2,5,6],3))
```

```python
#Better Approach
def merge(num1,m,num2,n):
    i=0
    j=0
    result=[]
    while i<m and j<n:
        if num1[i]<=num2[j]:
            result.append(num1[i])
            i+=1
        else:
            result.append(num2[j])
            j+=1
    while i<m:
        result.append(num1[i])
        i+=1
    while j<n:
        result.append(num2[j])
        j+=1
    for k in range(m+n):
        num1[k]=result[k]
    return num1
print(merge([1,2,3,0,0,0],3,[2,5,6],3))
```

```python
#Optimal Approach
def merge(num1,m,num2,n):
    i=m-1
    j=n-1
    k=m+n-1
    while i >=0 and j>=0:
        if num1[i]<num2[j]:
            num1[k]=num2[j]
            j-=1
        else:
            num1[k]=num1[i]
            i-=1
        k-=1
    while j>=0:
        num1[k]=num2[j]
        j-=1
        k-=1
print(merge([1,2,3,0,0,0],3,[2,5,6],3))
```

3Sum

```python
#Brute Force
def threeSum(nums):
    n=len(nums)
    result=[]
    seen=set()
    for i in range(n):
        for j in range(i+1,n):
            for k in range(j+1,n):
                if nums[i]+nums[j]+nums[k]==0:
                    triplet=sorted([nums[i],nums[j],nums[k]])
                    if tuple(triplet) not in seen:
                        seen.add(tuple(triplet))
                        result.append(triplet)
    return result
nums = [-1,0,1,2,-1,-4]
print(threeSum(nums))
```

```python
#Better Approach
def threeSum(nums):
    n=len(nums)
    result=set()
    for i in range(n):
        seen=set()
        for j in range(i+1,n):
            third=-(nums[i]+nums[j])
            if third in seen:
                triplet=tuple(sorted([nums,nums[j],third]))
                result.add(triplet)
            seen.add(nums[j])
    return [list(t) for t in result]
```
