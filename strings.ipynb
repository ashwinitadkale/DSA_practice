# 1.Remove Outermost Parentheses

Time Complexity: O(n), since we are performing a single traversal of the string.

Space Complexity: O(1), since we are using a few variables to track the current state.

```python
def Outermost_parentheses(st):
  result=""
  d=0
  for s in st:
    if s=="(":
      if d>0:
        result+=s
      d+=1
    else:
      d-=1
      if d>0:
        result+=s
  return result
st = "((()))"
print(Outermost_parentheses(st))
st1="(())))())"
print(Outermost_parentheses(st1))
```

# Reverse Words in a String

```python
def reverseWords(s):
    words=[]
    word=""
    for ch in s:
        if ch!=" ":
           word+=ch
        elif word:
             words.append(word)
             word=""
    if word:
       words.append(word)
    words.reverse()
    return " ".join(words)
s = "welcome to the jungle"
print(reverseWords(s))
```

```python
def reverseWords1(s):
    return " ".join(s.split()[::-1])
s = "welcome to the jungle"
print(reverseWords1(s))
```

# Longest Common Prefix

```python
def LoongestCommon(strs):
  strs.sort()
  first=strs[0]
  last=strs[-1]
  i=0
  while i<len(first) and first[i]==last[i]:
    i+=1
  return first[:i]
str = ["flower", "flow", "flight"]
print(LoongestCommon(str))
```

# Isomorphic String

```python
def isIsomorphic(s, t):
    s_set=set(s)
    t_set=set(t)
    if len(s_set)!=len(t_set):
        return False
    else:
        return True
s = "paper"
t="title"
print(isIsomorphic(s,t))
```
