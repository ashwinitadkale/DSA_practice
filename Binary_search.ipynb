# Search X in sorted array

```python
def binary_search_iterative(arr, target):
    left, right = 0, len(arr) - 1

    while left <= right:
        mid = (left + right) // 2

        if arr[mid] == target:
            return mid  # element found
        elif arr[mid] < target:
            left = mid + 1  # search right half
        else:
            right = mid - 1  # search left half

    return -1  # element not found
```

```python
def binary_search_recursive(arr, left, right, target):
    if left > right:
        return -1  # base case: not found

    mid = (left + right) // 2

    if arr[mid] == target:
        return mid
    elif arr[mid] < target:
        return binary_search_recursive(arr, mid + 1, right, target)
    else:
        return binary_search_recursive(arr, left, mid - 1, target)
```

# Implement Lower Bound

Brute force  

Time Complexity: O(N), where N = size of the given array.

Space Complexity: O(1), no extra space used.

```python
#brute force
def Lower_bound(arr,x):
  for i in range(len(arr)):
    if arr[i]>=x:
      return i
  return len(arr) #if x>all elements
arr = [3, 5, 8, 15, 19]
x = 9
print(Lower_bound(arr,x))
```

Optimal mathod

Time Complexity: O(logn), used for typical binary search

Space Complexity: O(1), no extra space used

```python
def lower_bound(arr,x):
  low,high=0,len(arr)-1
  ans=len(arr)
  while low<=high:
    mid=(low+high)//2
    if arr[mid]>=x:
      ans=mid
      high=mid-1
    else:
      low=mid+1
  return ans
arr = [3, 5, 8, 15, 19]
x = 9
print(lower_bound(arr,x))
```

# Floor and ceil in sorted array

Time Complexity:O(logN), where N = size of the given array. We are using the Binary Search algorithm

Space Complexity: O(1). No extra space used

```python
class FloorCeilFinder:
    def find_floor(self, arr, x):
        low, high = 0, len(arr) - 1
        ans = -1
        while low <= high:
            mid = (low + high) // 2
            if arr[mid] <= x:
                ans = arr[mid]  # Potential floor
                low = mid + 1
            else:
                high = mid - 1
        return ans

    def find_ceil(self, arr, x):
        low, high = 0, len(arr) - 1
        ans = -1
        while low <= high:
            mid = (low + high) // 2
            if arr[mid] >= x:
                ans = arr[mid]  # Potential ceil
                high = mid - 1
            else:
                low = mid + 1
        return ans

    def get_floor_and_ceil(self, arr, x):
        floor = self.find_floor(arr, x)
        ceil = self.find_ceil(arr, x)
        return floor, ceil

# Usage
arr = [3, 4, 4, 7, 8, 10]
x = 5
finder = FloorCeilFinder()
f, c = finder.get_floor_and_ceil(arr, x)
print(f"The floor and ceil are: {f} {c}")
```

```python
from _typeshed import AnyOrLiteralStr
def single_element(arr):
  n=len(arr)
  for i in range(n):
    if n==1:
      return AnyOrLiteralStr[0]
    for i in range(n):
      if i==0:
        if arr[i]!=arr[i+1]:
          return arr[i]
        elif i==n-1:
          if arr[i]!=arr[i-1]:
            return arr[i]
        else:
          if arr[i] != arr[i - 1] and arr[i] != arr[i + 1]:
            return arr[i]
  return -1

```

Peak element in Array

```python
#Brute force
def findPeakElement(nums):
    n=len(nums)
    for i in range(n):
        left=nums[i-1] if i>0 else float('-inf')
        right=nums[i+1] if i<n-1 else float('-inf')
        if nums[i]>left and nums[i]>right:
            return i 
    return -1
nums = [1,2,3,4,5,6,7,8,5,1]
index =findPeakElement(nums)
print(f"Peak at index: {index} with value: {nums[index]}")
```

```python
#Optimal Approach
def findPeak(nums):
    low,high=0,len(nums)-1
    while low<=high:
        mid=(low+high)//2
        if nums[mid]>nums[mid+1]:
            high=mid
        else:
            low=mid+1
    return low
nums = [1,2,3,4,5,6,7,8,5,1]
index =findPeak(nums)
print(f"Peak at index: {index} with value: {nums[index]}")
```

Minimum days to make M bouquets 

```python
# Brute Force 
def possibleFlowers(bloom_days,day,m,k):
    cnt=0
    no_of_bouquets=0
    for bloom in bloom_days:
        if bloom<=day:
            cnt+=1
            if cnt==k:
                no_of_bouquets+=1
                cnt=0
        else:
            cnt=0
    return no_of_bouquets>=m
def miniDays(arr,m,k):
    if (m*k)>len(arr):
        return -1
    for i in range(min(arr),max(arr)):
        if possibleFlowers(arr,i,m,k):
            return i
    return -1
bloom_days = [7, 7, 7, 7, 13, 11, 12, 7]
k = 3
m = 2
print(miniDays(bloom_days,m,k))
```

```python
#Optimal Approach
def minDays(bloomDay,m,k):
    if (m*k)>len(bloomDay):
        return -1
    def isPossible(day):
        cnt=0
        no_of_bouquets=0
        for bloom in bloom_days:
            if bloom<=day:
                cnt+=1
                if cnt==k:
                    no_of_bouquets+=1
                    cnt=0
            else:
                cnt=0
            if no_of_bouquets>=m:
                return True
        return False
    low=min(bloomDay)
    high=max(bloomDay)
    while low<high:
        mid=(low+high)//2
        if isPossible(mid):
            high=mid
        else:
            low=mid+1
    return low
bloom_days = [7, 7, 7, 7, 13, 11, 12, 7]
k = 3
m = 2
print(minDays(bloom_days,m,k))
```

Find the smallest Divisor Given a Threshold

```python
import math
def smallestDivisor(arr,limit):
    n=len(arr)
    max_val=max(arr)
    for d in range(1,max_val+1):
        total=0
        for num in arr:
            total+=math.ceil(num/d)
        if total<=limit:
            return d
    return -1
arr=[1,2,3,4,5]
limit=8
print(smallestDivisor(arr,limit))
```

```python
class Solution(object):
    def smallestDivisor(self, nums, threshold):
        
        def compute_sum(divisor):
            total = 0
            for n in nums:
                total += (n + divisor - 1) // divisor   # ✅ no float
            return total
        
        low, high = 1, max(nums)
        
        while low < high:
            mid = (low + high) // 2
            
            if compute_sum(mid) <= threshold:
                high = mid
            else:
                low = mid + 1
        
        return low
arr=[1,2,3,4,5]
limit=8
print(smallestDivisor(arr,limit))
```
