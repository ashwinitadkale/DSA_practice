Longest Consecutive sequence in array

```python
#Brute Force
def longestCons(nums):
    def ls(nums,x):
        for i in range(len(nums)):
            if nums[i]==x:
                return True
        return False
    longest=1
    for i in range(len(nums)):
        x=nums[i]
        cnt=1
        while ls(nums,x+1)==True:
            x=x+1
            cnt+=1
        longest=max(longest,cnt)
    return longest
nums=[100,4,200,1,3,2]
print(longestCons(nums))
```

```python
#Better Approach
def longestCons(nums):
    n=len(nums)
    if n==0:
        return 0
    nums.sort()
    lastSmaller=float('-inf')
    cnt=0
    longest=1
    for i in range(n):
        if nums[i]-1==lastSmaller:
            cnt+=1
            lastSmaller=nums[i]
        elif nums[i]!=lastSmaller:
            cnt=1
            lastSmaller=nums[i]
        longest=max(longest,cnt)
    return longest
a=[100,4,200,1,3,2]
print(longestCons(a))   
```

```python
#Optimal Approach
def longestCons(nums):
    n=len(nums)
    if n==0:
        return 0
    longest=1
    st=set(nums)
    for n in st:
        if (n-1) not in st:
            cnt=1
            while (n+cnt) in st:
                cnt+=1
            longest=max(cnt,longest)
            if longest>len(nums)/2:
                return longest
    return longest

#     for it in st:
#         if it-1 not in st:
#             cnt=1
#             x=it
#             while x+1 in st:
#                 x=x+1
#                 cnt=cnt+1
#             longest=max(longest,cnt)
#     return longest
a=[100,4,200,1,3,2]
print(longestCons(a))

                
        
```

Set matrix zeroes

```python
#Brute Force
def setMatrix(mat):
    n=len(mat)
    m=len(mat[0])
    for i in range(n):
        for j in range(m):
            if mat[i][j]==0:
                for col in range(m):
                    if mat[i][col]!=0:
                        mat[i][col]=-1
                for row in range(n):
                    if mat[row][j]!=0:
                        mat[row][j]=-1
    for i in range(n):
        for j in range(m):
            if mat[i][j]==-1:
                mat[i][j]=0
    return mat
print(setMatrix([[1,1,1],[1,0,1],[1,1,1]])) 
```

```python
#Better Approach
def setMatrix(mat):
    n=len(mat)
    m=len(mat[0])
    row=[0]*n
    col=[0]*m
    for i in range(n):
        for j in range(m):
            if mat[i][j]==0:
                row[i]=1
                col[j]=1
    for i in range(n):
        for j in range(m):
            if row[i]==1 or col[j]==1:
                mat[i][j]=0
    return mat
print(setMatrix([[1,1,1],[1,0,1],[1,1,1]])) 
```

Rotate Image by 90 degree

```python
#Brute Force 
def rotateImage(mat):
    n=len(mat)
    m=len(mat[0])
    for i in range(n):
        for j in range(i+1,n):
            mat[i][j],mat[j][i]=mat[j][i],mat[i][j]
        mat[i].reverse()
    return mat
print(rotateImage( [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]))    
```

TwoSum

```python
#Brute Force Approach #time complexity O(n2) #space O(1)
def twosum(nums,target):
    n=len(nums)
    for i in range(n):
        for j in range(i+1,n):
            if nums[i]+nums[j]==target:
                return [i,j]
nums = [2,7,11,15]
target = 9
print(twosum(nums,target))
```

```python
#Better Approach # time O(n log n) #space O(n)
def twosum(nums,target):
    arr=[]
    for i,num in enumerate(nums):
        arr.append((num,i))
    arr.sort()
    left=0
    right=len(arr)-1
    while left<right:
        total=arr[left][0]+arr[right][0]
        if total==target:
            return [arr[left][1],arr[right][1]]
        elif total<target:
            left+=1
        else:
            right-=1
nums = [2,7,11,15]
target = 9
print(twosum(nums,target))
```

```python
#Optimal Approach
def twosum(nums,target):
    d={}
    for i in range(len(nums)):
        complement=target-nums[i]
        if complement in d:
            return [d[complement],i]
        d[nums[i]]=i
nums = [2,7,11,15]
target = 9
print(twosum(nums,target))
```

Valid Anagram

```python
#using list # O(n2)
s="eat"
t="tea"
if len(s)!=len(t):
    print(False)
temp=list(t)
for ch in s:
    if ch in temp:
        temp.remove(ch)
    else:
        print(False)
print(True)
```

```python
#usinh HashMaps
s="anagram"
t="nagaram"
count={}
for ch in s:
    count[ch]=count.get(ch,0)+1
for ch in t:
    count[ch]=count.get(ch,0)-1
for val in count.values():
    if val!=0:
        print(False)
print(True)
```

GROUP ANAGRAMS

```python
#using Hashmap O(n log k)
strs=["eat","tea","tan","ate","nat","bat"]
hashmap={}
for word in strs:
    key=''.join(sorted(word))
    if key not in hashmap:
        hashmap[key]=[]
    hashmap[key].append(word)
print(list(hashmap.values()))
```
